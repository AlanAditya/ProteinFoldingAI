```python
def factorial(k):
    ans = 1
    for i in range(1, k+1):
        ans *= i
    return ans
```

```python
def C(n, r):
    if r > n:
        return 0
    if r == 0 or r == n:
        return 1
    num = 1
    for i in range(n, n-r, -1):
        num *= i
    den = 1
    for i in range(1, r+1):
        den *= i
    return num // den
```

```python
def S(n, k):
    ans = 0
    for r in range(0, k):
        print(f"C({k}, {r}) * {k-r}**({k}-{r}) * ((-1)**({r}))")
        ans += C(k, r) * (k-r)**(n) * ((-1)**(r))
    return ans // factorial(k)
```
